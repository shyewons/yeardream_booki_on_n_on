# 3차 실험 — 상위 조합 정확도 + 속도 재측정

2차 실험의 **14번 조합**(최종 적중@3 최고, 하이브리드 검색)은 느리다. 정확도가 비슷하다면 빠른 조합을 쓸 수 있는지 확인하려고 세 조합을 다시 돌린다.

|---|---|
| 13번 조합 | 형태소 질의 보강 · KURE-v1 · 벡터 서치만 · 구조+semantic 청킹 |
| 14번 조합 | 형태소 질의 보강 · bge-m3 · 하이브리드 검색 · 구조+semantic 청킹 |
| 15번 조합 (비교군) | kanana 쿼리 재작성 · KURE-v1 · 하이브리드 검색 · 구조+semantic 청킹 |

- 평가셋 · et5 · 형태소 판정 · 청크 · 채점(최종 적중@3)은 2차 실험과 같다.

- test에 더해 dev도 채점하고, 두 조합 사이의 차이를 McNemar 검정으로 확인한다.

```
# =====================================================================
# [셀 2] 공통 함수 — 정규화, 어절 단위 diff
# =====================================================================
PUNCT_RE = re.compile(r"[.,!?~…·;:\"'“”‘’()\[\]{}<>「」『』\-–—/]+")

def _nfc(s):
    return unicodedata.normalize("NFC", str(s)).strip()

def _core(tokens):
    return PUNCT_RE.sub("", " ".join(tokens))

def _sp(s):
    return re.sub(r"\s+", " ", PUNCT_RE.sub("", _nfc(s))).strip()

def _ns(s):
    return PUNCT_RE.sub("", _nfc(s)).replace(" ", "")

def extract_diffs(src, cor, ctx=CTX):
    """원문·교정문을 어절 단위로 비교. 문장 부호만 바뀐 조각은 제외"""
    a, b = _nfc(src).split(), _nfc(cor).split()
    out = []
    for op, i1, i2, j1, j2 in difflib.SequenceMatcher(None, a, b, autojunk=False).get_opcodes():
        if op == "equal":
            continue
        err_tok, cor_tok = a[i1:i2], b[j1:j2]
        if _core(err_tok) == _core(cor_tok):
            continue
        L, R = max(i1 - ctx, 0), min(i2 + ctx, len(a))
        Lb, Rb = max(j1 - ctx, 0), min(j2 + ctx, len(b))
        err_ctx = re.sub(r"[.!?~…]+$", "", " ".join(a[L:R]))
        cor_ctx = re.sub(r"[.!?~…]+$", "", " ".join(b[Lb:Rb]))
        out.append({"err": " ".join(err_tok), "cor": " ".join(cor_tok), "err_ctx": err_ctx, "cor_ctx": cor_ctx,
                    "query": f"{err_ctx} → {cor_ctx}", "i1": i1, "i2": i2})
    return out

for d in extract_diffs("오늘 친구 한명이 우리반에 왔다", "오늘 친구 한 명이 우리 반에 왔다"):
    print(d["query"])
```

```
# =====================================================================
# [셀 3] 평가셋 로드 — '사용' 문항 중 500개 층화 무작위 추출(1차 실험과 같은 500개)
# =====================================================================
df = pd.read_excel(EXCEL_PATH, sheet_name="검수", dtype=str)
df = df[df["사용 여부"].fillna("").str.strip() == "사용"].copy().reset_index(drop=True)
strata = df.groupby(["추출 그룹", "평가 용도"]).size()
quota = strata / strata.sum() * N_SAMPLE
alloc = np.floor(quota).astype(int)
for k in (quota - alloc).sort_values(ascending=False).index[:N_SAMPLE - alloc.sum()]:
    alloc[k] += 1
df = pd.concat([g.sample(n=alloc[k], random_state=SAMPLE_SEED) for k, g in df.groupby(["추출 그룹", "평가 용도"])]
              ).sort_values("문항ID").reset_index(drop=True)
df["gold"] = df.apply(lambda r: sorted({_nfc(x) for x in [r.get("정답 조항 1"), r.get("정답 조항 2")]
                                         if isinstance(x, str) and x.strip()}), axis=1)
df["src"] = df["오류 문장(맥락)"].map(_nfc)
print("평가 문항:", len(df), "| dev", (df["평가 용도"] == "dev").sum(), "/ test", (df["평가 용도"] == "test").sum())
```

```
# =====================================================================
# [셀 4] et5 교정 (Drive 캐시 재사용)
# =====================================================================
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, AutoModelForCausalLM

ET5_CACHE = f"{CACHE_DIR}/et5_{ET5_MODEL.split('/')[-1]}_b{ET5_BEAMS}_L{ET5_MAX_LEN}.json"
et5_cache = json.load(open(ET5_CACHE, encoding="utf-8")) if os.path.exists(ET5_CACHE) else {}
todo = [s for s in dict.fromkeys(df["src"]) if s not in et5_cache]
print(f"et5 교정 필요 {len(todo)}개 (캐시 {len(et5_cache)}개)")
if todo:
    et5_tok = AutoTokenizer.from_pretrained(ET5_MODEL)
    et5 = AutoModelForSeq2SeqLM.from_pretrained(ET5_MODEL).to(DEVICE).eval()
    for s in range(0, len(todo), ET5_BATCH):
        batch = todo[s:s + ET5_BATCH]
        enc = et5_tok([ET5_PREFIX + x for x in batch], return_tensors="pt", padding=True,
                      truncation=True, max_length=ET5_MAX_LEN).to(DEVICE)
        with torch.no_grad():
            gen = et5.generate(**enc, max_length=ET5_MAX_LEN, num_beams=ET5_BEAMS, early_stopping=True)
        for x, y in zip(batch, et5_tok.batch_decode(gen, skip_special_tokens=True)):
            et5_cache[x] = _nfc(y)
        json.dump(et5_cache, open(ET5_CACHE, "w", encoding="utf-8"), ensure_ascii=False)
        print(f"et5 교정 {min(s + ET5_BATCH, len(todo))}/{len(todo)}")
    del et5, et5_tok; gc.collect()
    if DEVICE == "cuda": torch.cuda.empty_cache()
df["et5"] = df["src"].map(et5_cache)
```

```
# =====================================================================
# [셀 5] diff(원문 vs et5) → 이 문항의 오류 조각 + et5 교정 상태
#   일치 = 정답대로 고침(주 평가 대상) / 다르게 = 다르게 고침 / 누락 = 안 고침
# =====================================================================
def locate(tokens, target):
    tgt = _sp(target).replace(" ", "")
    if not tgt:
        return None
    flat, owner = "", []
    for i, t in enumerate(tokens):
        c = PUNCT_RE.sub("", t); flat += c; owner += [i] * len(c)
    k = flat.find(tgt)
    return None if k < 0 else (owner[k], owner[k + len(tgt) - 1] + 1)

def match_segment(row):
    span = locate(row["src"].split(), row["틀린 말"])
    if span is None:
        return None, "위치 못 찾음"
    s, e = span
    over = [d for d in extract_diffs(row["src"], row["et5"])
            if (d["i1"] < e and d["i2"] > s) or (d["i1"] == d["i2"] and s <= d["i1"] <= e)]
    if not over:
        return None, "누락"
    gold_fix = _sp(row["고친 말"])
    for d in over:
        if gold_fix and gold_fix in _sp(d["cor_ctx"]):
            return d, "일치"
    return over[0], "다르게"

m_ = [match_segment(r) for _, r in df.iterrows()]
df["seg"] = [d for d, _ in m_]
df["et5_status"] = [s for _, s in m_]
df["query"] = [d["query"] if d else None for d in df["seg"]]
print("et5 교정 상태:", df["et5_status"].value_counts().to_dict())
```

```
# =====================================================================
# [셀 6] 맞춤법 본문 → 조항 분할(부록 '문장 부호' 제외)
# =====================================================================
FORCE_REDOWNLOAD = False
if FORCE_REDOWNLOAD or not os.path.exists(RULES_TXT):
    import requests
    from bs4 import BeautifulSoup
    soup = BeautifulSoup(requests.get(RULES_URL, timeout=30, headers={"User-Agent": "Mozilla/5.0"}).text, "html.parser")
    for t in soup(["script", "style", "nav", "header", "footer"]):
        t.decompose()
    for t in soup.find_all(["b", "strong", "em", "i", "u", "span", "font", "sup", "sub", "a", "mark"]):
        t.unwrap()
    for br in soup.find_all("br"):
        br.replace_with("\n")
    soup.smooth()
    open(RULES_TXT, "w", encoding="utf-8").write(re.sub(r"\n\s*\n+", "\n", soup.get_text("\n")))

raw = _nfc(open(RULES_TXT, encoding="utf-8").read())

def split_articles(text):
    marks, expect = [], 1
    for m in re.finditer(r"(?m)^\s*제\s*(\d+)\s*항", text):
        if int(m.group(1)) == expect:
            marks.append((expect, m.start())); expect += 1
        if expect > 57:
            break
    assert len(marks) >= 50, f"조항 경계를 {len(marks)}개만 찾았습니다."
    arts = [("맞춤법-0", text[:marks[0][1]])]
    for k, (n, st) in enumerate(marks):
        body = text[st: marks[k+1][1] if k + 1 < len(marks) else len(text)]
        m = re.search(r"(?m)^\s*\[?\s*부\s*록", body) if k + 1 == len(marks) else None
        if m:
            arts += [(f"맞춤법-{n}", body[:m.start()]), ("맞춤법-부록", body[m.start():])]
        else:
            arts.append((f"맞춤법-{n}", body))
    return [(lab, re.sub(r"[ \t]+", " ", t).strip()) for lab, t in arts if t.strip()]

ARTICLES = [(l, t) for l, t in split_articles(raw) if l != "맞춤법-부록"]
print("조항 수:", len(ARTICLES), "(정상 58) | 총 글자:", sum(len(t) for _, t in ARTICLES),
      "| 예문 띄어쓰기 정상:", "아는 것이 힘이다" in raw and "아는 것 이 힘이다" not in raw)
```

```
# =====================================================================
# [셀 7] 청킹 2종 + 대표 조항 라벨
#   고정500: 500자씩 / 구조+semantic: 조항 경계 + 하위 10% 유사도 지점, 200~800자
# =====================================================================
from sentence_transformers import SentenceTransformer, CrossEncoder

def chunk_fixed(articles, size=FIXED_LEN):
    full, owner = "", []
    for lab, t in articles:
        full += t + "\n"; owner += [lab] * (len(t) + 1)
    chunks = []
    for s in range(0, len(full), size):
        txt = full[s:s+size].strip()
        if txt:
            cnt = Counter(owner[s:s+size])
            chunks.append({"text": txt, "label": cnt.most_common(1)[0][0], "labels": sorted(cnt)})
    return chunks

def _sentences(t):
    return [p.strip() for p in re.split(r"(?<=[.!?])\s+|\n+", t) if p.strip()]

def chunk_struct_semantic(articles, model, pct=SEM_PCT, mn=SEM_MIN, mx=SEM_MAX):
    per_art = []
    for lab, t in articles:
        sents = _sentences(t)
        if len(sents) >= 2:
            E = model.encode(sents, normalize_embeddings=True, batch_size=32, show_progress_bar=False)
            sim = (E[:-1] * E[1:]).sum(1)
        else:
            sim = np.array([])
        per_art.append((lab, sents, sim))
    thr = np.percentile(np.concatenate([s for _, _, s in per_art if len(s)]), pct)
    chunks = []
    for lab, sents, sim in per_art:
        cur, cur_len = [], 0
        def flush():
            nonlocal cur, cur_len
            if cur:
                chunks.append({"text": " ".join(cur), "label": lab, "labels": [lab]})
            cur, cur_len = [], 0
        for i, s in enumerate(sents):
            while len(s) > mx:
                flush(); chunks.append({"text": s[:mx], "label": lab, "labels": [lab]}); s = s[mx:]
            if cur and cur_len + len(s) > mx:
                flush()
            cur.append(s); cur_len += len(s) + 1
            if i < len(sim) and sim[i] <= thr and cur_len >= mn:
                flush()
        if cur:
            if chunks and chunks[-1]["label"] == lab and cur_len < mn and len(chunks[-1]["text"]) + cur_len + 1 <= mx:
                chunks[-1]["text"] += " " + " ".join(cur)
            else:
                flush()
    return chunks

_sem = SentenceTransformer(SEM_SPLIT_MODEL, device=DEVICE)
if DEVICE == "cuda": _sem.half()
CHUNKS = {"고정500": chunk_fixed(ARTICLES), "구조+semantic": chunk_struct_semantic(ARTICLES, _sem)}
del _sem; gc.collect()
if DEVICE == "cuda": torch.cuda.empty_cache()
for k, v in CHUNKS.items():
    L = np.array([len(c["text"]) for c in v])
    print(f"{k}: {len(v)}개, 평균 {L.mean():.0f}자, 최대 {L.max()}자, 조항 걸침 {sum(len(c['labels']) > 1 for c in v)}개")
```

```
# =====================================================================
# [셀 8] 형태소 분석 판정 — diff 조각의 문법 특징 추출 → 판정 유형(①~④) + 질의 보강 문자열
#   ① 표기만 / ② 띄어쓰기 + 구체 범주(의존 명사·단위 명사·보조 용언·조사·열거) / ③ 띄어쓰기 + 표기 섞임 / ④ 일반 띄어쓰기만
#   규칙 기반(LLM 아님). 조항 번호는 쓰지 않는다.
# =====================================================================
from kiwipiepy import Kiwi
kiwi = Kiwi()

_UNIT = set("개 명 번 살 년 월 일 시 분 초 시간 달 주 개월 배 권 장 대 마리 잔 병 가지 군데 층 학년 차 회 벌 채 켤레 강 판 그루 송이 척 쪽 원 킬로 미터 센티".split())
_NUM_MM = set("한 두 세 네 다섯 여섯 일곱 여덟 아홉 열 몇 스무 서 너 여러 수십 수백 첫".split())
_LIST = set("겸 내지 대 등 및 등등 등속 등지".split())
_JONG = " ㄱㄲㄳㄴㄵㄶㄷㄹㄺㄻㄼㄽㄾㄿㅀㅁㅂㅄㅅㅆㅇㅈㅊㅋㅌㅍㅎ"

def _bounds(s):
    s = PUNCT_RE.sub("", _nfc(s)); b, k = set(), 0
    for ch in s:
        if ch == " ": b.add(k)
        else: k += 1
    return b

def _toks(s):
    s = PUNCT_RE.sub("", _nfc(s)); m, k = [], 0
    for ch in s:
        m.append(k)
        if ch != " ": k += 1
    m.append(k)
    return [(m[t.start], t.form, t.tag) for t in kiwi.tokenize(s)]

def _jong(ch):
    o = ord(ch) - 0xAC00
    return _JONG[o % 28] if 0 <= o < 11172 else ""

def _shift_typo(x, y):
    if len(x) != 1 or len(y) != 1: return False
    ox, oy = ord(x) - 0xAC00, ord(y) - 0xAC00
    if not (0 <= ox < 11172 and 0 <= oy < 11172): return False
    ax, ay = (ox // 588, (ox % 588) // 28, ox % 28), (oy // 588, (oy % 588) // 28, oy % 28)
    d = [i for i in range(3) if ax[i] != ay[i]]
    if len(d) != 1: return False
    i = d[0]; p = tuple(sorted((ax[i], ay[i])))
    return ((i == 0 and p in {(9, 10), (3, 4), (7, 8), (0, 1), (12, 13)}) or
            (i == 2 and p in {(19, 20), (1, 2)}) or (i == 1 and p in {(1, 3), (5, 7)}))

def gram_features(err, cor):
    """반환: (띄어쓰기 특징 목록, 표기 특징 목록)"""
    sp_f, lt_f = [], []
    e, c = _ns(err), _ns(cor)
    sm = difflib.SequenceMatcher(None, e, c, autojunk=False)
    c2e = {b + t: a + t for a, b, size in sm.get_matching_blocks() for t in range(size)}
    be, bc = _bounds(err), _bounds(cor)
    T = _toks(cor); start = {}
    for i, t in enumerate(T): start.setdefault(t[0], i)
    for k in sorted(bc):                                    # 새로 띄운 경계
        if k in c2e and c2e[k] in be: continue
        i = start.get(k)
        if i is None: continue
        _, f, tg = T[i]; p = T[i-1] if i > 0 else None
        if f in _LIST: sp_f.append(f"열거하는 말 '{f}' 띄어쓰기")
        elif (tg == "NNB" or f in _UNIT) and p and (p[2] in ("SN", "NR") or p[1] in _NUM_MM):
            sp_f.append(f"수 뒤 단위 명사 '{f}' 띄어쓰기")
        elif tg == "NNB" or f in ("거", "걸", "건", "게"): sp_f.append(f"의존 명사 '{f}' 띄어쓰기")
        elif tg == "VX": sp_f.append(f"보조 용언 '{f}' 띄어쓰기")
        elif tg.startswith("J"): sp_f.append(f"조사 '{f}' 앞 띄어쓰기")
        elif tg.startswith("XS"): sp_f.append(f"접미사 '{f}' 앞 띄어쓰기")
        else: sp_f.append(f"단어 '{f}' 앞 띄어쓰기")
    inv = {v: k for k, v in c2e.items()}
    for j in sorted(be):                                    # 붙인 경계
        k = inv.get(j)
        if k is None or k in bc: continue
        i = start.get(k)
        if i is None: continue
        _, f, tg = T[i]
        if tg.startswith("J") or tg == "VCP": sp_f.append(f"조사 '{f}' 붙여 쓰기")
        elif tg.startswith("XS") or tg.startswith("E"): sp_f.append(f"접미사·어미 '{f}' 붙여 쓰기")
        elif tg == "VX": sp_f.append(f"보조 용언 '{f}' 붙여 쓰기")
        else: sp_f.append(f"'{f}' 앞 붙여 쓰기(한 단어)")
    for op, a1, a2, b1, b2 in sm.get_opcodes():             # 글자 표기 변화
        if op == "equal": continue
        x, y = e[a1:a2], c[b1:b2]
        if _shift_typo(x, y):
            lt_f.append(f"'{x}'를 '{y}'로 고친 표기(입력 오타)"); continue
        wy = c[max(b1-2, 0):b2+2]
        if ("돼" in x + y or "됬" in x or "됀" in x): lt_f.append("'되'와 '돼' 구별, '되어'가 줄어든 '돼' 표기")
        elif "않" in y or ("안" in y and "않" in x): lt_f.append("'않다'(아니하다)와 부사 '안' 구별 표기")
        elif "잖" in y or "찮" in y: lt_f.append("'-지 않-'이 줄어든 '-잖-', '-하지 않-'이 줄어든 '-찮-' 표기")
        elif any(s in x for s in ("께", "껄", "쑤록", "찌")) and any(s in y for s in ("게", "걸", "수록", "지")):
            lt_f.append(f"어미 '{y}' 예사소리 표기(-ㄹ게, -ㄹ걸, -ㄹ수록, -ㄹ지)")
        elif x == "가" and y == "까" and b1 > 0 and _jong(c[b1-1]) == "ㄹ" and c[b2:b2+1] != "지":
            lt_f.append("의문 어미 '-ㄹ까' 된소리 표기")
        elif ("히" in y and "이" in x) or ("이" in y and "히" in x): lt_f.append(f"부사 끝 '-이'와 '-히' 구별 표기('{wy}')")
        elif ("더" in y and "드" in x) or ("던" in y and "든" in x) or ("든" in y and "던" in x):
            lt_f.append("지난 일 '-더라, -던'과 선택 '-든' 구별 표기")
        elif "로서" in wy or "로써" in wy: lt_f.append("'로서'와 '로써' 구별 표기")
        elif "의" in y and "이" in x: lt_f.append(f"'ㅢ' 표기('{wy}')")
        elif "며칠" in wy: lt_f.append("'며칠' 표기")
        else:
            stem = next((t for t in kiwi.tokenize(c[max(b1-3, 0):b2+3]) if t.tag in ("VV", "VA", "VX")), None)
            lt_f.append(f"용언 '{stem.form}다'의 어간과 어미를 밝혀 적은 표기('{wy}')" if stem else f"'{x}'를 '{y}'로 고친 표기")
    return list(dict.fromkeys(sp_f)), list(dict.fromkeys(lt_f))

SPACE_SPECIFIC = ("의존 명사", "수 뒤 단위 명사", "보조 용언", "조사 '", "열거하는 말")   # 제5장 구체 규칙

def decide(sp_f, lt_f):
    """판정 유형: ①표기만 ②띄어쓰기+구체 ③띄어쓰기+표기 ④일반 띄어쓰기"""
    if not sp_f:
        return "①표기만"
    if any(k in f for f in sp_f for k in SPACE_SPECIFIC):
        return "②띄어쓰기+구체"
    if lt_f:
        return "③띄어쓰기+표기"
    return "④일반 띄어쓰기"

feat = [gram_features(d["err"], d["cor"]) if d else ([], []) for d in df["seg"]]
df["sp_feats"] = [s for s, _ in feat]
df["lt_feats"] = [l for _, l in feat]
df["decide"] = [decide(s, l) if d else None for (s, l), d in zip(feat, df["seg"])]
# 형태소 질의 보강: 특징(최대 4개, 입력 오타 제외) + 원래 diff 질의
df["query_enr"] = [("; ".join([f for f in (s + l) if "입력 오타" not in f][:4]) + " | " if (s or l) else "") + q
                   if isinstance(q, str) else None for (s, l), q in zip(feat, df["query"])]
print("판정 유형:", df["decide"].value_counts().to_dict())
for _, r in df[df["query"].notna()].head(4).iterrows():
    print(f"- {r['decide']} | {r['query_enr'][:120]}")
```

```
# =====================================================================
# [셀 9] kanana 재작성(베이스라인 쿼리) — 1차 실험과 같은 캐시 파일을 다시 씀
# =====================================================================
_rw_key = REWRITE_SYSTEM + REWRITE_PROMPT + json.dumps(REWRITE_FEWSHOT, ensure_ascii=False)
REWRITE_CACHE = f"{CACHE_DIR}/rewrite_{REWRITE_MODEL.split('/')[-1]}_{hashlib.md5(_rw_key.encode()).hexdigest()[:8]}.json"
rw_cache = json.load(open(REWRITE_CACHE, encoding="utf-8")) if os.path.exists(REWRITE_CACHE) else {}
todo = [q for q in dict.fromkeys(df["query"].dropna()) if q not in rw_cache]
print(f"재작성 필요 {len(todo)}개 (캐시 {len(rw_cache)}개)")
ART_RE = re.compile(r"(한글\s*맞춤법\s*)?제\s*\d+\s*항(\s*의?\s*\d+)?(에\s*따르면|에\s*따라|에서는|에\s*의하면)?[\s,]*")

def clean_rw(o, q):
    o = o.strip().split("\n")[0].strip().strip('"').strip("'")
    o = re.sub(r"^(검색어|출력)\s*[:：]\s*", "", o)
    return ART_RE.sub("", o).strip(" ,.") or q

if todo:
    tok = AutoTokenizer.from_pretrained(REWRITE_MODEL); tok.padding_side = "left"
    if tok.pad_token is None:
        tok.pad_token = tok.eos_token
    llm = AutoModelForCausalLM.from_pretrained(
        REWRITE_MODEL, torch_dtype=torch.float16 if DEVICE == "cuda" else torch.float32).to(DEVICE).eval()
    def _prompt(q):
        msgs = [{"role": "system", "content": REWRITE_SYSTEM}]
        for x, y in REWRITE_FEWSHOT:
            msgs += [{"role": "user", "content": REWRITE_PROMPT.format(query=x)}, {"role": "assistant", "content": y}]
        msgs.append({"role": "user", "content": REWRITE_PROMPT.format(query=q)})
        return tok.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
    for s in range(0, len(todo), REWRITE_BATCH):
        batch = todo[s:s + REWRITE_BATCH]
        enc = tok([_prompt(q) for q in batch], return_tensors="pt", padding=True).to(DEVICE)
        with torch.no_grad():
            gen = llm.generate(**enc, max_new_tokens=REWRITE_MAX_NEW_TOKENS, do_sample=False,
                               pad_token_id=tok.pad_token_id, eos_token_id=tok.eos_token_id)
        for q, o in zip(batch, tok.batch_decode(gen[:, enc["input_ids"].shape[1]:], skip_special_tokens=True)):
            rw_cache[q] = {"raw": o.strip(), "clean": clean_rw(o, q)}
        json.dump(rw_cache, open(REWRITE_CACHE, "w", encoding="utf-8"), ensure_ascii=False)
        print(f"재작성 {min(s + REWRITE_BATCH, len(todo))}/{len(todo)}")
    del llm, tok; gc.collect()
    if DEVICE == "cuda": torch.cuda.empty_cache()

def _rw(q):
    if not isinstance(q, str): return None
    v = rw_cache.get(q)
    return (v["clean"] if isinstance(v, dict) else v) or q
df["query_rw"] = df["query"].map(_rw)
```

```
# =====================================================================
# [셀 10] 모델 로드 + 벡터DB + BM25 + 검색 함수 (한 번만 실행)
# =====================================================================
from rank_bm25 import BM25Okapi

def ktok(s):
    return [t.form for t in kiwi.tokenize(s.replace("→", " ")) if not t.tag.startswith("S") or t.tag in ("SL", "SN", "SH")]

if "MODELS" not in globals():
    MODELS = {}
for name, path in EMBED_MODELS.items():
    if name not in MODELS:
        m = SentenceTransformer(path, device=DEVICE)
        if DEVICE == "cuda": m.half()
        MODELS[name] = m
if "reranker" not in MODELS:
    rr = CrossEncoder(RERANK_MODEL, max_length=512, device=DEVICE)
    if DEVICE == "cuda": rr.model.half()
    MODELS["reranker"] = rr

CHUNK_EMB, BM25 = {}, {}
for ck in CHUNKS:
    for emb in EMBED_MODELS:
        CHUNK_EMB[(emb, ck)] = MODELS[emb].encode([c["text"] for c in CHUNKS[ck]], normalize_embeddings=True,
                                                 batch_size=16, show_progress_bar=False,
                                                 convert_to_numpy=True).astype(np.float32)
    BM25[ck] = BM25Okapi([ktok(c["text"]) for c in CHUNKS[ck]])

def search_all(queries, emb, chunker, method):
    Q = MODELS[emb].encode(list(queries), normalize_embeddings=True, batch_size=32,
                           show_progress_bar=False, convert_to_numpy=True).astype(np.float32)
    S = Q @ CHUNK_EMB[(emb, chunker)].T
    if method == "벡터만":
        return [[int(i) for i in np.argsort(-s)[:max(TOPK_EVAL, 10)]] for s in S]
    cand = []
    for qi, q in enumerate(queries):
        rrf = defaultdict(float)
        for r, i in enumerate(np.argsort(-S[qi])[:VEC_TOPN]): rrf[int(i)] += 1 / (RRF_K + r + 1)
        for r, i in enumerate(np.argsort(-BM25[chunker].get_scores(ktok(q)))[:BM25_TOPN]): rrf[int(i)] += 1 / (RRF_K + r + 1)
        cand.append([i for i, _ in sorted(rrf.items(), key=lambda x: -x[1])[:RRF_TOPN]])
    texts = [c["text"] for c in CHUNKS[chunker]]
    scores = MODELS["reranker"].predict([(q, texts[i]) for q, cl in zip(queries, cand) for i in cl],
                                        batch_size=64, show_progress_bar=False)
    scores = np.nan_to_num(np.asarray(scores, dtype=np.float32), nan=-1e9)
    out, p = [], 0
    for cl in cand:
        sc = scores[p:p+len(cl)]; p += len(cl)
        out.append([cl[j] for j in np.argsort(-sc)])
    return out

has_q = df["query"].notna().values
q_idx = np.where(has_q)[0]

def run_search(qcol, emb, chunker, method):
    """검색 결과를 Drive에 저장. 같은 조건이면 저장본 재사용(중간에 끊겨도 이어서 실행)"""
    sig = _md5({"q": df.loc[has_q, ["문항ID", qcol]].values.tolist(), "emb": EMBED_MODELS[emb], "method": method,
                "chunks": _md5([c["text"] for c in CHUNKS[chunker]]), "p": [VEC_TOPN, BM25_TOPN, RRF_K, RRF_TOPN, RERANK_MODEL]})
    path = f"{RANK_DIR}/{sig}.json"; t = time.time()
    if os.path.exists(path):
        rq, msg = json.load(open(path)), "저장된 결과"
    else:
        rq = search_all(df.loc[has_q, qcol].tolist(), emb, chunker, method)
        json.dump(rq, open(path, "w")); msg = f"{time.time()-t:.0f}s"
    ranks = [None] * len(df)
    for i, rk in zip(q_idx, rq): ranks[i] = rk
    return ranks, msg

for emb in EMBED_MODELS:                                   # 스모크 테스트
    for method in ["벡터만", "하이브리드+리랭킹"]:
        r = search_all(df["query_enr"].dropna().head(3).tolist(), emb, "구조+semantic", method)
        print(f"테스트 OK: {emb} | {method} → {[CHUNKS['구조+semantic'][i]['label'] for i in r[0][:3]]}")
if DEVICE == "cuda":
    print(f"GPU 메모리: {torch.cuda.memory_allocated()/1e9:.1f}GB")
```

```
# =====================================================================
# [셀 11] RAG 3개 조합(#13 · #14 · #15)만 실행 — 정확도용 검색 결과 + 걸리는 시간
#   평가셋·et5·판정·청크·kanana 재작성은 셀 0~10 그대로(16개 조합 평가와 같은 데이터)
#   조합 번호는 16개 조합 때와 같다. 검색은 저장본을 쓰지 않고 새로 돌려 시간을 잰다
#   et5 교정·형태소 판정 시간은 세 조합에 공통이라 빼고, 질의 → 검색 결과까지만 잰다
# =====================================================================
BASE_CFG = {"쿼리": "kanana 재작성", "임베딩": "bge-m3", "검색": "벡터만", "청킹": "고정500"}
ALT = {"쿼리": "형태소 질의 보강", "임베딩": "KURE", "검색": "하이브리드+리랭킹", "청킹": "구조+semantic"}
QCOL = {"kanana 재작성": "query_rw", "형태소 질의 보강": "query_enr"}
AXES = list(BASE_CFG)
ORDER = [(), ("쿼리",), ("임베딩",), ("검색",), ("청킹",),
         ("쿼리", "임베딩"), ("쿼리", "검색"), ("쿼리", "청킹"), ("임베딩", "검색"), ("임베딩", "청킹"), ("검색", "청킹"),
         ("쿼리", "임베딩", "검색"), ("쿼리", "임베딩", "청킹"), ("쿼리", "검색", "청킹"), ("임베딩", "검색", "청킹"),
         ("쿼리", "임베딩", "검색", "청킹")]
COMBOS = []
for n, changed in enumerate(ORDER, 1):
    cfg = {a: (ALT[a] if a in changed else BASE_CFG[a]) for a in AXES}
    cfg.update({"#": n, "바꾼 것": ", ".join(changed) or "없음(베이스라인)"})
    COMBOS.append(cfg)

PICK = [13, 14, 15]
N_REPEAT = 3            # 전체 질의 배치 시간을 몇 번 재서 중앙값을 쓸지
N_SINGLE = 100          # 서비스처럼 한 문장씩 넣어 잴 질의 수
MEASURE_KANANA = False  # True면 #15의 kanana 재작성 시간도 잰다(모델 4.6GB를 다시 올림, 약 3~5분)

COMBOS = [c for c in COMBOS if c["#"] in PICK]
cfg_of = {c["#"]: c for c in COMBOS}
for c in COMBOS:
    print(f"#{c['#']}: {c['쿼리']} · {c['임베딩']} · {c['검색']} · {c['청킹']}")

def _sync():
    if DEVICE == "cuda":
        torch.cuda.synchronize()

def _saved_ranks(qcol, emb, chunker, method):
    """16개 조합 평가 때 Drive에 저장한 검색 결과(있으면). 새로 돌린 결과와 같은지 확인용"""
    sig = _md5({"q": df.loc[has_q, ["문항ID", qcol]].values.tolist(), "emb": EMBED_MODELS[emb], "method": method,
                "chunks": _md5([c["text"] for c in CHUNKS[chunker]]), "p": [VEC_TOPN, BM25_TOPN, RRF_K, RRF_TOPN, RERANK_MODEL]})
    path = f"{RANK_DIR}/{sig}.json"
    return json.load(open(path)) if os.path.exists(path) else None

q_list = {k: df.loc[has_q, QCOL[cfg_of[k]["쿼리"]]].tolist() for k in PICK}
print(f"\n검색 질의 수: {len(q_list[PICK[0]])}개 (평가 500문항 중 et5가 고친 곳이 있는 문항)")

# 1) 전체 질의 배치 검색 — 정확도용 결과 + 시간
for k in PICK:                                             # 워밍업(첫 호출의 GPU 초기화 시간 제외)
    c = cfg_of[k]
    search_all(q_list[k][:8], c["임베딩"], c["청킹"], c["검색"])

RANKS, T_BATCH = {}, {}
for k in PICK:
    c = cfg_of[k]; ts = []
    for _ in range(N_REPEAT):
        _sync(); t0 = time.perf_counter()
        rq = search_all(q_list[k], c["임베딩"], c["청킹"], c["검색"])
        _sync(); ts.append(time.perf_counter() - t0)
    ranks = [None] * len(df)
    for i, rk in zip(q_idx, rq):
        ranks[i] = rk
    RANKS[k], T_BATCH[k] = ranks, float(np.median(ts))
    saved = _saved_ranks(QCOL[c["쿼리"]], c["임베딩"], c["청킹"], c["검색"])
    if saved is None:
        note = "16개 조합 저장본 없음"
    else:
        labs = [ch["label"] for ch in CHUNKS[c["청킹"]]]
        same = np.mean([[labs[j] for j in a[:3]] == [labs[j] for j in b[:3]] for a, b in zip(rq, saved)])
        note = f"16개 조합 저장본과 top3 일치 {same:.1%}"
    print(f"#{k} 배치 {T_BATCH[k]:.1f}s (반복 {N_REPEAT}회 중앙값) | {note}")

# 2) 한 문장씩 검색 — 서비스에서 실제로 기다리는 시간
rng = np.random.default_rng(0)
single_idx = rng.choice(len(q_list[PICK[0]]), size=min(N_SINGLE, len(q_list[PICK[0]])), replace=False)
T_SINGLE = {}
for k in PICK:
    c = cfg_of[k]; ts = []
    for j in single_idx:                                  # 세 조합 모두 같은 문항으로 잰다
        _sync(); t0 = time.perf_counter()
        search_all([q_list[k][j]], c["임베딩"], c["청킹"], c["검색"])
        _sync(); ts.append(time.perf_counter() - t0)
    T_SINGLE[k] = np.array(ts) * 1000
    print(f"#{k} 한 문장 검색: 중앙값 {np.median(T_SINGLE[k]):.0f}ms · 95% {np.percentile(T_SINGLE[k], 95):.0f}ms")

# 3) (선택) #15의 kanana 재작성 시간 — #15는 검색 전에 sLLM 생성이 한 번 더 들어간다
T_KANANA = None
if MEASURE_KANANA and 15 in PICK:
    _tok = AutoTokenizer.from_pretrained(REWRITE_MODEL); _tok.padding_side = "left"
    if _tok.pad_token is None:
        _tok.pad_token = _tok.eos_token
    _llm = AutoModelForCausalLM.from_pretrained(
        REWRITE_MODEL, torch_dtype=torch.float16 if DEVICE == "cuda" else torch.float32).to(DEVICE).eval()
    def _rw_prompt(q):
        msgs = [{"role": "system", "content": REWRITE_SYSTEM}]
        for x, y in REWRITE_FEWSHOT:
            msgs += [{"role": "user", "content": REWRITE_PROMPT.format(query=x)}, {"role": "assistant", "content": y}]
        msgs.append({"role": "user", "content": REWRITE_PROMPT.format(query=q)})
        return _tok.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
    raw_q = df.loc[has_q, "query"].tolist()
    ts = []
    for n_, j in enumerate(single_idx[:21]):
        enc = _tok([_rw_prompt(raw_q[j])], return_tensors="pt").to(DEVICE)
        _sync(); t0 = time.perf_counter()
        with torch.no_grad():
            _llm.generate(**enc, max_new_tokens=REWRITE_MAX_NEW_TOKENS, do_sample=False,
                          pad_token_id=_tok.pad_token_id, eos_token_id=_tok.eos_token_id)
        _sync()
        if n_ > 0:                                          # 첫 번째는 워밍업
            ts.append(time.perf_counter() - t0)
    T_KANANA = np.array(ts) * 1000
    print(f"#15 kanana 재작성(한 문장): 중앙값 {np.median(T_KANANA):.0f}ms · 95% {np.percentile(T_KANANA, 95):.0f}ms")
    del _llm, _tok; gc.collect()
    if DEVICE == "cuda": torch.cuda.empty_cache()
```

```
# =====================================================================
# [셀 12] 근거 결정기 + 지표 — 16개 조합 평가(2차 실험)와 같은 채점
#   최종 적중@3 = 판정 유형에 따라 [제2항] + RAG 결과를 합친 근거 목록(최대 3개)에 정답 조항이 있는가
#   채점 문항 = et5 '일치' 문항(test 250 · dev 114). test+dev = 둘을 합친 364문항
# =====================================================================
def resolve_evidence(kind, rag_labels):
    """서비스에서 그대로 쓰는 함수: 판정 유형 + 검색 결과(대표 조항, 순위순) → 최종 근거 목록"""
    rag = rag_labels or []
    if kind == "①표기만":
        return rag[:3]
    if kind == "④일반 띄어쓰기":
        return ["맞춤법-2"]
    return ["맞춤법-2"] + [l for l in rag if l != "맞춤법-2"][:2]

def _mean(x):
    return float(np.mean(x)) if len(x) else np.nan

def mcnemar_p(a, b):
    n01, n10 = int(((a == 1) & (b == 0)).sum()), int(((a == 0) & (b == 1)).sum())
    n = n01 + n10
    return 1.0 if n == 0 else min(1.0, 2 * sum(comb(n, i) for i in range(min(n01, n10) + 1)) / 2 ** n)

def boot_ci(x, B=2000, seed=0):
    if len(x) == 0: return (np.nan, np.nan)
    m = x[np.random.default_rng(seed).integers(0, len(x), size=(B, len(x)))].mean(1)
    return float(np.percentile(m, 2.5)), float(np.percentile(m, 97.5))

def boot_diff(a, b, B=5000, seed=0):
    idx = np.random.default_rng(seed).integers(0, len(a), size=(B, len(a)))
    d = a[idx].mean(1) - b[idx].mean(1)
    return float(np.percentile(d, 2.5)), float(np.percentile(d, 97.5))

def scores_for(ranks, chunker):
    labs = [c["label"] for c in CHUNKS[chunker]]
    rag1, rag3, mrr, fin = [], [], [], []
    for rk, gold, kind in zip(ranks, df["gold"], df["decide"]):
        if rk is None:
            rag1.append(0); rag3.append(0); mrr.append(0.0); fin.append(0); continue
        top = [labs[i] for i in rk]
        hit = next((p for p, l in enumerate(top[:TOPK_EVAL]) if l in gold), None)
        rag1.append(int(hit == 0)); rag3.append(int(hit is not None)); mrr.append(0.0 if hit is None else 1 / (hit + 1))
        fin.append(int(any(l in gold for l in resolve_evidence(kind, top))))
    return np.array(rag1), np.array(rag3), np.array(mrr), np.array(fin)

is_dev = (df["평가 용도"] == "dev").values
ok = (df["et5_status"] == "일치").values
test_ok, dev_ok = ~is_dev & ok, is_dev & ok
has2 = df["gold"].apply(lambda g: "맞춤법-2" in g).values

SC = {k: scores_for(RANKS[k], cfg_of[k]["청킹"]) for k in PICK}
rows = []
for k in PICK:
    r1, r3, mrr, fin = SC[k]
    lo, hi = boot_ci(fin[test_ok])
    q_ms = np.median(T_SINGLE[k]) + (np.median(T_KANANA) if (k == 15 and T_KANANA is not None) else 0)
    rows.append({"#": k, "구성": f"{cfg_of[k]['쿼리']} · {cfg_of[k]['임베딩']} · {cfg_of[k]['검색']}",
                 "test 최종": fin[test_ok].mean(), "test 95%CI": f"{lo:.3f}~{hi:.3f}",
                 "test 제2항 정답": fin[test_ok & has2].mean(), "test 제2항 아닌": fin[test_ok & ~has2].mean(),
                 "test RAG R@1": r1[test_ok].mean(), "test RAG R@3": r3[test_ok].mean(),
                 "dev 최종": fin[dev_ok].mean(), "test+dev 최종": fin[ok].mean(),
                 "전체 성공률@3": (fin * ok)[~is_dev].mean(),
                 "배치 시간(s)": T_BATCH[k], "한 문장 검색(ms)": np.median(T_SINGLE[k]), "한 문장 질의+검색(ms)": q_ms})
res = pd.DataFrame(rows).set_index("#")
print(f"채점 문항: test {int(test_ok.sum())}개(제2항 정답 {int((test_ok & has2).sum())}) · dev {int(dev_ok.sum())}개 · 합계 {int(ok.sum())}개")
fmt = {c: "{:.3f}" for c in res.columns if c not in ("구성", "test 95%CI")}
fmt.update({"배치 시간(s)": "{:.1f}", "한 문장 검색(ms)": "{:.0f}", "한 문장 질의+검색(ms)": "{:.0f}"})
_show(res, fmt)
if T_KANANA is None:
    print("※ #15의 '한 문장 질의+검색'에는 kanana 재작성 시간이 빠져 있습니다. 재려면 셀 11에서 MEASURE_KANANA = True")

# 둘씩 비교 — 같은 문항에서 맞음·틀림이 갈린 경우로 판단. 차이 95%CI가 0을 걸치면 '구별 안 됨'
pair_rows = []
for a_, b_ in [(13, 14), (13, 15), (14, 15)]:
    fa, fb = SC[a_][3], SC[b_][3]
    for name, mask in [("test", test_ok), ("dev", dev_ok), ("test+dev", ok)]:
        x, y = fa[mask], fb[mask]
        lo, hi = boot_diff(x, y)
        pair_rows.append({"비교": f"#{a_} − #{b_}", "평가셋": name, "문항": int(mask.sum()),
                          "차이": x.mean() - y.mean(), "차이 95%CI": f"{lo:+.3f}~{hi:+.3f}",
                          "앞만 맞음": int(((x == 1) & (y == 0)).sum()), "뒤만 맞음": int(((x == 0) & (y == 1)).sum()),
                          "p값": mcnemar_p(x, y),
                          "판단": "구별 안 됨" if (lo <= 0 <= hi) else ("앞이 높음" if lo > 0 else "뒤가 높음")})
pair_tbl = pd.DataFrame(pair_rows)
print("\n[둘씩 비교] 차이 = 앞 조합 − 뒤 조합 (최종 적중@3)")
_show(pair_tbl.set_index(["비교", "평가셋"]), {"차이": "{:+.3f}", "p값": "{:.3f}"})

# 판정 유형별 — 세 조합 나란히
kd = df["decide"].fillna("질의 없음").values
kind_rows = []
for kk in ["①표기만", "②띄어쓰기+구체", "③띄어쓰기+표기", "④일반 띄어쓰기"]:
    m = test_ok & (kd == kk)
    kind_rows.append({"판정 유형": kk, "test 문항 수": int(m.sum()),
                      **{f"#{k} 최종": _mean(SC[k][3][m]) for k in PICK}})
kind_tbl = pd.DataFrame(kind_rows).set_index("판정 유형")
print("\n[판정 유형별 test 최종 적중@3]  ④는 검색을 안 쓰므로 세 조합이 같다")
_show(kind_tbl, {"test 문항 수": "{:.0f}", **{f"#{k} 최종": "{:.3f}" for k in PICK}})

# #13과 #14가 갈린 test 문항 — AI허브 문장이 나오므로 이 출력은 저장소·문서에 붙이지 않는다
lab13 = [ch["label"] for ch in CHUNKS[cfg_of[13]["청킹"]]]
lab14 = [ch["label"] for ch in CHUNKS[cfg_of[14]["청킹"]]]
f13, f14 = SC[13][3], SC[14][3]
diff_i = np.where(test_ok & (f13 != f14))[0]
print(f"\n[#13과 #14가 갈린 test 문항] {len(diff_i)}개 (#13만 맞음 {int((f13[diff_i] == 1).sum())} · #14만 맞음 {int((f14[diff_i] == 1).sum())})")
for i in diff_i:
    r = df.iloc[i]
    t13 = [lab13[j] for j in RANKS[13][i][:3]]; t14 = [lab14[j] for j in RANKS[14][i][:3]]
    print(f"- {r['문항ID']} | {r['decide']} | 정답 {r['gold']} | {'#13' if f13[i] else '#14'}만 맞음"
          f"\n    #13 근거 {resolve_evidence(r['decide'], t13)} · #14 근거 {resolve_evidence(r['decide'], t14)}"
          f"\n    {r['틀린 말']} → {r['고친 말']}")

# 요약
gap = res.loc[14, "한 문장 검색(ms)"] - res.loc[13, "한 문장 검색(ms)"]
print("\n[요약] #13을 쓰면 #14 대비")
for name in ["test", "dev", "test+dev"]:
    p_ = pair_tbl[(pair_tbl["비교"] == "#13 − #14") & (pair_tbl["평가셋"] == name)].iloc[0]
    print(f"  {name:>8}: 최종 적중 {p_['차이']:+.3f} (95%CI {p_['차이 95%CI']}, p={p_['p값']:.3f}) → {p_['판단']}")
print(f"  한 문장당 검색 시간 {-gap:+.0f}ms")
```

```
# =====================================================================
# [셀 13] 저장 — 결과표·둘씩 비교·판정 유형별·문항별 상세. detail 파일에는 AI허브 문장이 있으니 저장소에 올리지 않는다
# =====================================================================
stamp = time.strftime("%Y%m%d_%H%M")
res.to_csv(f"{OUT_DIR}/h2_rag3_summary_{stamp}.csv", encoding="utf-8-sig")
pair_tbl.to_csv(f"{OUT_DIR}/h2_rag3_pairs_{stamp}.csv", index=False, encoding="utf-8-sig")
kind_tbl.to_csv(f"{OUT_DIR}/h2_rag3_decide_types_{stamp}.csv", encoding="utf-8-sig")
detail = df[["문항ID", "평가 용도", "추출 그룹", "src", "et5", "틀린 말", "고친 말", "et5_status",
             "decide", "query", "query_rw", "query_enr", "gold"]].copy()
for k in PICK:
    labs = [c["label"] for c in CHUNKS[cfg_of[k]["청킹"]]]
    tops = [[labs[i] for i in rk] if rk else [] for rk in RANKS[k]]
    detail[f"#{k} RAG top3"] = [", ".join(t[:3]) for t in tops]
    detail[f"#{k} 최종 근거"] = [", ".join(resolve_evidence(kk, t)) if t else "" for kk, t in zip(df["decide"], tops)]
detail.to_csv(f"{OUT_DIR}/h2_rag3_detail_{stamp}.csv", index=False, encoding="utf-8-sig")
print("저장:", OUT_DIR)
```
